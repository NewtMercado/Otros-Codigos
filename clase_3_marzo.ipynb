```python
def f(x):
    return x**2-1

a = 0 # Limite inferior
b = 2 # Limite superior
tol = 1e-5 # Tolerancia
max_iter = 100000 # Maximo numero de iteraciones
p = (a+b) / 2 # Punto medio inicial

iter = 0 # Inicializacion del numero de iteraciones

while (iter < max_iter and abs(f(p)) > tol):
    if (f(a)*f(p) < 0):
        b = p
    else:
        a = p

    p = (a+b) / 2 # Actualizacion del punto medio
    iter += 1

print(iter,p,f(p))
```

```python
m = 10

for n in range(1,m+1):
    impares = []
    for k in range(n):
        impares.append(2*k+1)

    suma_impares = sum(impares)

    while suma_impares != n**3:
        k += 1
        impares.pop(0)
        impares.append(2*k+1)
        suma_impares = sum(impares)

    print("n = " + str(n) + ":",impares)
```
