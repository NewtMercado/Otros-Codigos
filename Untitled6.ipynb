```python
import numpy as np

# Ingresar los valores de x y y por teclado
x_values = input("Ingrese los valores del vector empaquetado x separados por espacios: ").split()
y_values = input("Ingrese los valores del vector disperso y separados por espacios (índice:valor): ").split()

# Convertir los valores de x a un vector empaquetado
x = np.array(x_values, dtype=int)

# Convertir los valores de y a un vector disperso (diccionario de índices y valores)
y = {}
for item in y_values:
    index, value = map(int, item.split(":"))
    y[index] = value

# Función para calcular el producto punto entre un vector empaquetado y un vector disperso
def sparse_dot_product(x, y):
    dot_product = 0
    for index, value in y.items():
        dot_product += x[index-1] * value  # Restamos 1 al índice porque Python usa índices base 0
    return dot_product

# Calcular el producto punto
result = sparse_dot_product(x, y)

print("Producto punto:", result)
```
