# Vector

```python
import numpy as np


# Definimos usando la clase spaesevector para guardar el vector empaquetado

class SparseVector:
    def __init__(self, nz, value, index):
        self.nz = nz
        self.value = value
        self.index = index

# Definimos la función para empaquetar un vector disperso

def gatherVector(v):
    nz = np.count_nonzero(v)
    indices = np.nonzero(v)[0]
    values = v[indices]
    return SparseVector(nz, values, indices)

# usamos la funcion para hallar el producto punto de los vectores empaquetados

def sparse_dot_product(packed_vector, ypacked_vector):
    result = 0
    for i in range(ypacked_vector.nz):
        index = ypacked_vector.index[i]
        value = ypacked_vector.value[i]
        for j in range(packed_vector.nz):
            if packed_vector.index[j] == index:
                result += value * packed_vector.value[j]
    return result

# utilizamos la función para hallar el producto punto de un vector empaquetado y un vector disperso

def dot_product(packed_vector, vector):
    result = 0
    for i in range(packed_vector.nz):
        index = packed_vector.index[i]
        value = packed_vector.value[i]
        result += value * vector[index]
    return result

# Definir la función para multiplicar un vector por un escalar utilizando scalar_multiply

def scalar_multiply(vector, scalar):
    nz = vector.nz
    indices = vector.index
    values = vector.value * scalar
    return SparseVector(nz, values, indices)

# sumamos dos vectores empaquetados

def add_vectors(packed_vector1, packed_vector2):
    indices1 = packed_vector1.index
    values1 = packed_vector1.value
    indices2 = packed_vector2.index
    values2 = packed_vector2.value

    indices = np.union1d(indices1, indices2)
    values = np.zeros_like(indices)

    for i, index in enumerate(indices):
        value1 = values1[indices1 == index]
        value2 = values2[indices2 == index]
        values[i] = value1.sum() + value2.sum()

    nz = np.count_nonzero(values)
    return SparseVector(nz, values, indices)

# utilizamos las funciones para hallar las normas de un vector empaquetado

def norm_1(vector):
    return np.sum(np.abs(vector))

def norm_2(vector):
    return np.sqrt(np.sum(vector ** 2))

def norm_infinity(vector):
    return np.max(np.abs(vector))



#definimos los vectores pidiento por teclado
#x es el vector empaquetado y y es el disperso
# Empaquetar los vectores de entrada
y = np.array([0,0,0,0,30,0,11,0,0,57])
x = np.array([0,0,67,0,0,19,0,86,0,14])


x_sparse = gatherVector(x)
y_sparse = gatherVector(y)


# pedimos por teclado al usuario la opcion que desea
option = int(input("Selecciona una opción\n"
                   "1. Producto punto entre dos vectores empaquetados\n"
                   "2. Producto punto entre un vector empaquetado y uno disperso\n"
                   "3. Suma de dos vectores empaquetados\n"
                   "4. Norma 1\n"
                   "5. Norma 2\n"
                   "6. Norma infinito\n"))

if option == 1:
    # Multiplicar dos vectores empaquetados
    print("x_empaquetado:")
    print("Número de elementos no nulos:", x_sparse.nz)
    print("Valores:", x_sparse.value)
    print("Indices:", x_sparse.index)

    print("\ny_empaquetado:")
    print("Número de elementos no nulos:", y_sparse.nz)
    print("Valores:", y_sparse.value)
    print("Indices:", y_sparse.index)
    result = sparse_dot_product(x_sparse, y_sparse)
    print("Producto punto (dos vectores empaquetados):", result)
elif option == 2:
    # Producto punto entre vector empaquetado y vector sin empaquetar
    print("x_empaquetado:")
    print("Número de elementos no nulos:", x_sparse.nz)
    print("Valores:", x_sparse.value)
    print("Indices:", x_sparse.index)

    print("\ny_empaquetado:")
    print("Número de elementos no nulos:", y_sparse.nz)
    print("Valores:", y_sparse.value)
    print("Indices:", y_sparse.index)
    dense_result = dot_product(x_sparse, y)
    print("Producto punto (vector empaquetado y vector sin empaquetar):", dense_result)
elif option == 3:
    # Sumar dos vectores empaquetados
    print("x_empaquetado:")
    print("Número de elementos no nulos:", x_sparse.nz)
    print("Valores:", x_sparse.value)
    print("Indices:", x_sparse.index)

    print("\ny_empaquetado:")
    print("Número de elementos no nulos:", y_sparse.nz)
    print("Valores:", y_sparse.value)
    print("Indices:", y_sparse.index)
    scalar = float(input("Digite el número escalar para escalar el segundo vector empaquetado: "))
    y_scaled = scalar_multiply(y_sparse, scalar)
    sum_result = add_vectors(x_sparse, y_scaled)
    print("Suma de dos vectores empaquetados:")
    print("Indices:", sum_result.index)
    print("Valores:", sum_result.value)
    print("Número de elementos no nulos:", sum_result.nz)
elif option == 4:
    # Calcular la norma 1 de un vector empaquetado
    vec_to=int(input("Digite 1: Vector x  2: Vector y  "))
    if vec_to==1:
        norm1 = norm_1(x_sparse.value)
        print("Norma 1 del primer vector empaquetado:", norm1)
    elif vec_to==2:
        norm1 = norm_1(y_sparse.value)
        print("Norma 1 del segundo vector empaquetado:", norm1)
elif option == 5:
    # Calcular la norma 2 de un vector empaquetado
    vec_to=int(input("Digite 1: Vector x  2: Vector y  "))
    if vec_to==1:
        norm2 = norm_2(x_sparse.value)
        print("Norma 2 del primer vector empaquetado:", norm2)
    elif vec_to==2:
        norm2 = norm_2(y_sparse.value)
        print("Norma 2 del segundo vector empaquetado:", norm2)
elif option == 6:
    # Calcular la norma infinito de un vector empaquetado
    vec_to=int(input("Digite 1: Vector x  2: Vector y  "))
    if vec_to==1:
        norm_inf = norm_infinity(x_sparse.value)
        print("Norma infinito del primer vector empaquetado:", norm_inf)
    elif vec_to==2:
        norm_inf = norm_infinity(y_sparse.value)
        print("Norma infinito del primer vector empaquetado:", norm_inf)
else:
    print("Opción inválida.")




```

# Matriz

```python
import math

class Matrix:
    def __init__(self, rows, cols):
        self.rows = rows
        self.cols = cols
        self.data = [[0.0] * cols for _ in range(rows)]  #Inicializar la matriz con ceros

    def asignar(self, row, col, value):
        self.data[row][col] = value  #Asignar un valor a una ubicación específica en la matriz

    def disperso(self, filename):
        with open(filename, 'w') as file:
            file.write(f"{self.rows} {self.cols}\n")  #Escribir las dimensiones de la matriz en el archivo
            for row in self.data:
                file.write(" ".join(str(element) for element in row))  # Escribir los elementos de la matriz fila por fila
                file.write("\n")  #Agregar un salto de línea después de cada fila

    def densa(self, filename):
        with open(filename, 'w') as file:
            file.write(f"{self.rows} {self.cols}\n")  #Escribir las dimensiones de la matriz en el archivo
            for i in range(self.rows):
                for j in range(self.cols):
                    value = self.data[i][j]
                    if value != 0.0:
                        file.write(f"{i+1} {j+1} {value}\n")  #Escribir las coordenadas y valores de los elementos no cero

    def frobenius(self):
        norm = 0.0
        for i in range(self.rows):
            for j in range(self.cols):
                norm += self.data[i][j] ** 2  #Sumar los cuadrados de los elementos de la matriz
        norm = math.sqrt(norm)  #Calcular la raíz cuadrada de la suma
        return norm

def leer_matr(filename):
    with open(filename, 'r') as file:
        dimensions = file.readline().split()
        rows = int(dimensions[0])
        cols = int(dimensions[1])
        matrix = Matrix(rows, cols)  #Crear una nueva matriz con las dimensiones leídas del archivo
        for line in file:
            values = line.split()
            row = int(values[0]) - 1
            col = int(values[1]) - 1
            value = float(values[2])
            matrix.asignar(row, col, value)  #Agregar elementos a la matriz
        return matrix


def suma_matr(matrix1, matrix2):
    if matrix1.rows != matrix2.rows or matrix1.cols != matrix2.cols:
        print("Las matrices no tienen el mismo orden.")
        exit()
    result = Matrix(matrix1.rows, matrix1.cols)  #Crear una nueva matriz del mismo tamaño
    for i in range(matrix1.rows):
        for j in range(matrix1.cols):
            result.data[i][j] = matrix1.data[i][j] + matrix2.data[i][j]  #Sumar los elementos correspondientes
    return result


def transpuesta(matrix):
    result = Matrix(matrix.cols, matrix.rows)  #Crear una nueva matriz con las dimensiones invertidas
    for i in range(matrix.rows):
        for j in range(matrix.cols):
            result.data[j][i] = matrix.data[i][j]  #Intercambiar filas por columnas
    return result



#Lee las matrices desde los archivos de texto
matrix1 = leer_matr('matrix1.txt')
matrix2 = leer_matr('matrix2.txt')

#Pregunta al usuario si desea sumar o sacar la transpuesta
opc = input("¿Deseas sumar las matrices o sacar la transpuesta? (sumar,transpuesta,norma): ")

if opc == "sumar":
    #Suma las matrices
    result_matrix = suma_matr(matrix1, matrix2)
    #Guarda la matriz resultante en un archivo dispersa
    result_matrix.disperso('matriz_disperso.txt')

    #Guarda la matriz resultante en un archivo densa
    result_matrix.densa('matriz_densa.txt')
elif opc == "transpuesta":
    #Transpuesta de una matriz
    cual = int(input("Cual de las matrices quiere sacar la transpuesta: "))
    if cual==1:
      transpuesta1 = transpuesta(matrix1)
      result_matrix = transpuesta1
      #Guarda la matriz transpuesta en un archivo dispersa
      transpuesta1.disperso('transpuesta_disperso.txt')

      #Guarda la matriz transpuesta en un archivo densa
      transpuesta1.densa('transpuesta_densa.txt')

    elif cual ==2:
      transpuesta2 = transpuesta(matrix2)
      result_matrix = transpuesta2
      #Guarda la matriz transpuesta en un archivo dispersa
      transpuesta2.disperso('transpuesta_disperso.txt')

      #Guarda la matriz transpuesta en un archivo densa
      transpuesta2.densa('transpuesta_densa.txt')

elif opc== "norma":
  matr = input("Cual de las matriz quieres calcular la norma de Frobenius? ")

  if matr == "1":
      #Calcula la norma de Frobenius de la matriz 1
      norm_frobenius = matrix1.frobenius()
      print("Norma de Frobenius:", norm_frobenius)
  elif matr == "2":
      #Calcula la norma de Frobenius de la matriz 2
      norm_frobenius = matrix2.frobenius()
      print("Norma de Frobenius:", norm_frobenius)
  else:
      print("Seleccione una opcion correcta")
      exit()


else:
    print("Seleccione una opcion correcta")
    exit()
```
