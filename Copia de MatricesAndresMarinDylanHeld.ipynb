```python
## presentado por Andres Marin y Dylan Held
## Este programa lo realizamos para sumar, restar, multiplicar dos matrices, trasponer una matriz
## mostrar una matriz señalando cuales son los puntos sillas
import numpy as np
## pedimos por teclado que escoja la opcion que desea
print("""Hola, seleccione una opcion
1: Sumar dos matrices
2: Restar dos matrices
3: Multiplicar dos matrices
4: Transponer una matriz
5: Mostrar una matriz señalando cuales son los puntos de silla""")
opc=int(input())
## definimos las filas y columnas con enteros
m = int(input("digite el número de filas m para la matriz: "))
n = int(input("digite el número de columnas n para la matriz: "))
## m=filas y n= columnas
## definimos los matrices
matriz1 = np.zeros((m,n))
matriz2 = np.zeros((m,n))
matriz3 = np.zeros((m,n))
print ("Ingrese la primer matriz: ")
##Pedimos por teclado el rango de las matriz 1 y matriz 2 respectivamente
for i in range(m):
		for j in range(n):
			matriz1[i][j] = int(input("A (%d,%d): " % (i, j)))
print ("Ingrese la segunda matriz: ")
for i in range(n):
		for j in range(m):
			matriz2[i][j] = int(input("B (%d,%d): " % (i, j)))
## Empezamos con cada una de las opciones que tenemos
## EL codigo de la suma de 2 matrices
if opc == 1:
	for i in range(m):
		for j in range(n):
				matriz3[i][j] += matriz1[i][j] + matriz2[i][j]
	print("La suma de las matrices es:",matriz3)
## El codigo de la resta de 2 matrices
elif opc == 2:
	for i in range(m):
		for j in range(n):
				matriz3[i][j] += matriz1[i][j] - matriz2[i][j]
	print ("la resta de las matrices es:", matriz3)
## multiplicamos 2 matrices
elif opc == 3:
	for i in range(m):
			for j in range(n):
				for k in range(m):
					matriz3[i][j] += matriz1[i][k] * matriz2[k][j]

	print("la multiplicacion de las matrices es: ",matriz3)
# en esta opcion pedimos por teclado cual de las dos matrices desea hallarle la transpuesta
elif opc == 4:
	print("De que matriz desea la transpuesta matriz 1 o matriz 2")
	u=int(input())
	if u==1:
			for i in range(m):
					for j in range(n):
							matriz3[i][j] = matriz1[j][i]
	else:
			for i in range(m):
					for j in range(n):
							matriz3[i][j] = matriz2[j][i]
	print("la traspuesta de la matriz es:",matriz3)
 # En esta opcion buscamos los puntos de silla de la matriz
else:
  # Preguntamos a cual de las dos matrices quiere que busquemos los puntos de silla de la matriz
  print("De que matriz desea encontrar los puntos de silla de la matriz 1 o matriz 2")
  u=int(input())
  # Calculamos los puntos de silla de la matriz depediendo de cual de las dos matrices quiera buscarlo
  if u==1:
    f=0
    for i in range(m):
      min=matriz1[i][0]
      c=0
      for j in range(n):
        if matriz1[i][j]<min:
          min=matriz1[i][j]
          c=j
      max=matriz1[0][c]
      for k in range (m):
        if matriz1[k][c]>max:
          max=matriz1[k][c]
      if min==max:
        print("El punto de silla es: ",min)
        print("El lugar del punto de silla en la matriz es: "+str(i)+","+str(c));
        f=1
    if f==0:
      print(" la matriz no tiene punto de silla")
  else:
    f=0
    for i in range(m):
      min=matriz2[i][0]
      c=0
      for j in range(n):
        if matriz2[i][j]<min:
          min=matriz2[i][j]
          c=j
      max=matriz2[0][c]
      for k in range (m):
        if matriz2[k][c]>max:
          max=matriz2[k][c]
      if min==max:
        print("El punto de silla es: ",min)
        print("El lugar del punto de silla en la matriz es: "+str(i)+","+str(c));
        f=1
    if f==0:
      print("La matriz no tiene punto de silla")





```
