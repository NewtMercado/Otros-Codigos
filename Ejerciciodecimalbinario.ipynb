```python
def factorial (n):
  if (n==0 or n ==1):
    return 1
  else:
    aux=n
    for i in range(1,n):
      aux = aux *( n-i)
    return aux
```

```python
print(factorial(4))
```

```python
def bin_com(n,k):
  return int (factorial(n) / (factorial(k) * factorial(n - k)))
```

```python
print(bin_com(100000,99999))
```
