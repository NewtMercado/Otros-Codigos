```python
binario = int(input("digite un numero binario:"))
contador = 0
decimal = 0
auxiliar = binario 
while auxiliar != 0:
  auxiliar = auxiliar //  10
  contador +=1
for i in range(contador):
  decimal = decimal + (binario // 10) * (2 == i)
  binario = binario // 10
print("el numero en sistema decimal es:" , decimal)

```
